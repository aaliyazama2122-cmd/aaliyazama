```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


data = {
    "Hours_Studied": [2, 5, 3, 7, 1, 6, 4, 8, 5, 25],
    "Attendance": [60, 90, 75, 95, 50, 88, 80, 98, 85, 70],
    "Previous_Score": [45, 70, 55, 82, 40, 78, 65, 90, 72, 58],
    "Final_Score": [42, 76, 58, 88, 35, 82, 68, 94, 79, 60]
}

df = pd.DataFrame(data)


print("First Five Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)


print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


Q1 = df["Hours_Studied"].quantile(0.25)
Q3 = df["Hours_Studied"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Hours_Studied"] < lower_limit) |
    (df["Hours_Studied"] > upper_limit)
]

print("\nUnusual Values (Outliers):")
print(outliers["Hours_Studied"])


plt.figure(figsize=(8, 5))
sns.boxplot(y=df["Hours_Studied"])

plt.title("Box Plot of Student Study Hours")
plt.ylabel("Hours Studied")
plt.show()


print("\nThree Observations:")
print("1. Most students studied between 1 and 8 hours.")
print("2. One unusual value of 25 hours was identified.")
print("3. The dataset contains no missing values.")
```
